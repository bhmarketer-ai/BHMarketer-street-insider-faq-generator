# BHMarketer Street Insider FAQ Generator: Search Walkthrough

This notebook runs the [BHMarketer Street Insider FAQ Generator](https://github.com/bhmarketer-ai/BHMarketer-street-insider-faq-generator) on 35 real-world questions about buying a press release on Street Insider, checks each one finds the expected FAQ, and generates the FAQ with FAQPage schema. Built by [BHMarketer.ai](https://bhmarketer.ai/street-insider-press-releases/).

## 1. Load the tool and the datasets

```python
import os, sys
import pandas as pd

BASES = ["/kaggle/input/street-insider-faq-dataset", "..", "."]
base = next(b for b in BASES if os.path.exists(os.path.join(b, "street_insider_faq.py")))
data = base if os.path.exists(os.path.join(base, "street_insider_faqs.csv")) else os.path.join(base, "dataset")
sys.path.insert(0, base)
import street_insider_faq as si

faqs = pd.read_csv(os.path.join(data, "street_insider_faqs.csv"))
bench = pd.read_csv(os.path.join(data, "street_insider_search_benchmarks.csv"))
print(len(faqs), "FAQs |", len(bench), "benchmark questions")
```

## 2. FAQs per category

```python
faqs.groupby("category").size().rename("faqs").to_frame()
```

## 3. Run every benchmark question

```python
rows = []
for r in bench.itertuples():
    a = si.ask(r.query)
    rows.append({"query": r.query, "found": a["id"], "expected": r.expected_id,
                 "confidence": a["confidence"], "band": a["band"], "pass": a["id"] == r.expected_id})
results = pd.DataFrame(rows)
print(f"{results['pass'].sum()}/{len(results)} benchmark questions matched")
results
```

```python
results.groupby("band").size().rename("questions").to_frame()
```

## 4. Ask your own question

```python
for r in si.search("how quickly does a crypto release go live", 3):
    print(f"Q{r['id']} [{r['confidence']}] {r['question']}")

best = si.ask("how much does a street insider press release cost")
print("\n", best["answer"])
```

## 5. Generate the FAQ with FAQPage schema

```python
print(si.export("jsonld", "pricing")[:1200])
```
